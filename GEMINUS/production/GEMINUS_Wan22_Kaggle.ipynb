```python
from pathlib import Path
import base64
import hashlib
import io
import json
import os
import shutil
import subprocess
import sys
import time
from zipfile import ZipFile

import requests

# Use "sample" to render S01 only, or "full" for the complete film.
MODE = "full"
WIDTH, HEIGHT = 384, 672
MODEL_FRAMES = 65
FPS, SHOT_SECONDS = 24, 4
OUT = Path("/kaggle/working")
TEMP = Path("/kaggle/temp/geminus")
TEMP.mkdir(parents=True, exist_ok=True)
CLIPS = OUT / "GEMINUS_clips"
CLIPS.mkdir(parents=True, exist_ok=True)
COMFY = TEMP / "ComfyUI"
FRAMES = TEMP / "frames"
FRAMES.mkdir(exist_ok=True)
SERVER_OUTPUT = TEMP / "comfy_output"
SERVER_OUTPUT.mkdir(exist_ok=True)
LOG = OUT / "GEMINUS_render.log"
ROOT_URL = "http://127.0.0.1:8188"
CHECKPOINT = "Wan2.2-I2V-A14B"
QUANT = "Q4_K_S"
ENCODER = "umt5-xxl-encoder-Q4_K_M.gguf"
VAE = "wan_2.1_vae.safetensors"
PROMPTS = {1: 'Very slow dolly toward the upper body of the ivory anatomical mannequin. Keep the sculpture still and preserve its proportions. Soft ivory museum lighting, a subtle parallax shift, no text.', 2: 'Slow, shallow camera arc across the layered head and neck anatomy. Reveal the eyes, ears, optic and neck structures through the existing layers. Keep all anatomy stable and non-gory.', 3: 'Hold the paired cerebral hemispheres, then use a clean soft cross-dissolve to the transparent medial temporal view. Keep the brain shapes intact. Let soft light draw attention to the paired internal structures.', 4: 'Slow camera glide down the torso from the two asymmetrical lungs to the kidneys and adrenal glands. Keep anatomical geometry in place and avoid stretching or adding organs. Warm diffused light.', 5: 'Begin on the deep hip muscle display. Pull back, then reveal two independent inset exhibits with a gentle editorial wipe. The hip stays in its own upper view, the two reproductive studies stay separate below. No biological morphing.', 6: 'Slow vertical pullback along the stationary musculoskeletal exhibit, from shoulder girdle and upper arms to both hands, thighs, knees and feet. Keep the limb anatomy stable and the framing symmetrical.', 7: 'Smooth macro push along a DNA double helix as the strands locally unzip at a replication fork. Reveal daughter duplexes gradually. Warm champagne depth and controlled scientific visualization.', 8: 'Drift closer to the replication fork, then use a clear time-compressing optical dissolve to a compact replicated chromosome. Show two sister chromatids joined at one central constriction. Hold the final composition.', 9: 'In a simplified mitotic cell, move the separated chromatids toward opposite spindle poles, then show the membrane pinching and the appearance of two distinct daughter cells, one nucleus each. Stay at the same warm microscopic scale.', 10: 'A restrained camera push across the two sisters from neck to waist. The sister on the left steadies a stethoscope and the sister on the right makes a small cinema lens adjustment. Keep fingers, clothes and props stable. No faces.', 11: 'Track a highlight on the curved stethoscope tube, then match cut to two separate glossy black sculptural curves in warm ivory space. Keep the two pieces distinct as they drift closer.', 12: 'Bring two glossy black rounded curves into alignment with the exact asymmetric reference heart. Slow their movement and settle without a wobble. Hold the final heart shape on ivory.', 13: 'Locked camera. Keep the supplied logo artwork, wordmark and spacing pixel-stable. Fade in only the tagline, then hold on the final card. This fade can be composed directly in the edit.'}
NEGATIVE = ("low quality, flicker, shaky camera, jitter, anatomy distortion, extra limbs, "
            "oversaturated colors, neon, gore, skin wounds, watermark, captions, cartoon")
SHOTS = {number: (f"S{number:02d}_START.jpg", f"S{number:02d}_END.jpg") for number in range(1, 14)}

def command(*args, **kwargs):
    return subprocess.run([str(arg) for arg in args], check=True, **kwargs)

def log(message):
    print(message, flush=True)
    with LOG.open("a", encoding="utf-8") as handle:
        print(message, file=handle)

def unpack_assets(archive_bytes):
    with ZipFile(io.BytesIO(archive_bytes)) as archive:
        for first, last in SHOTS.values():
            for name in (first, last):
                (FRAMES / name).write_bytes(archive.read(f"frames/{name}"))

def ffmpeg(*args):
    command("ffmpeg", "-nostdin", "-y", "-hide_banner", "-loglevel", "error", "-filter_threads", "2", "-filter_complex_threads", "2", *args)

def validate_clip(path, seconds):
    data = json.loads(command("ffprobe", "-v", "error", "-show_streams", "-show_format", "-of", "json", path, capture_output=True, text=True).stdout)
    video = next(stream for stream in data["streams"] if stream["codec_type"] == "video")
    if (video["width"], video["height"]) != (720, 1280) or int(video["nb_frames"]) != seconds * FPS:
        raise ValueError(f"Wrong size or frame count in {path.name}")
    if abs(float(data["format"]["duration"]) - seconds) > 0.025:
        raise ValueError(f"Wrong duration in {path.name}")

def final_card(target):
    first, last = (FRAMES / name for name in SHOTS[13])
    ffmpeg("-loop", "1", "-framerate", FPS, "-i", first,
           "-loop", "1", "-framerate", FPS, "-i", last,
           "-filter_complex", "[0:v]scale=720:1280,setsar=1,format=yuv420p[a];"
           "[1:v]scale=720:1280,setsar=1,format=yuv420p[b];"
           "[a][b]blend=all_expr='A*(1-min(1,T/0.65))+B*min(1,T/0.65)'[v]",
           "-map", "[v]", "-frames:v", FPS * SHOT_SECONDS, "-an", "-c:v", "libx264",
           "-preset", "fast", "-threads", "2", "-crf", "16", "-movflags", "+faststart", target)
    validate_clip(target, SHOT_SECONDS)

def prepare_comfy():
    import torch
    if not torch.cuda.is_available():
        raise RuntimeError("Enable GPU T4 x2 in Kaggle Settings, then restart and run.")
    log(f"GPU: {torch.cuda.get_device_name(0)} | PyTorch {torch.__version__}")
    if shutil.disk_usage(TEMP).free < 25 * 1024 ** 3 and not (COMFY / "models" / "diffusion_models" / expert("High")).is_file():
        raise RuntimeError("At least 25 GiB of free temporary storage is required for the initial setup.")
    if not (COMFY / "main.py").is_file():
        command("git", "clone", "--depth", "1", "https://github.com/Comfy-Org/ComfyUI.git", COMFY)
    gguf_node = COMFY / "custom_nodes" / "ComfyUI-GGUF"
    if not (gguf_node / "nodes.py").is_file():
        command("git", "clone", "--depth", "1", "https://github.com/city96/ComfyUI-GGUF.git", gguf_node)
    stamp = TEMP / "deps.ok"
    if not stamp.exists():
        command(sys.executable, "-m", "pip", "install", "-q",
                "-r", COMFY / "requirements.txt", "-r", gguf_node / "requirements.txt", "huggingface_hub")
        stamp.touch()
    from huggingface_hub import hf_hub_download

    def fetch(repo, remote, model_dir):
        target = COMFY / "models" / model_dir / Path(remote).name
        if target.is_file():
            return
        target.parent.mkdir(parents=True, exist_ok=True)
        log(f"Downloading {target.name}")
        local = Path(hf_hub_download(repo_id=repo, filename=remote, local_dir=TEMP / "weights"))
        local.replace(target)

    pack = "Comfy-Org/Wan-2.2"
    quantized = "QuantStack/Wan2.2-I2V-A14B-GGUF"
    for stage in ("High", "Low"):
        fetch(quantized, f"{stage}Noise/{expert(stage)}", "diffusion_models")
        fetch(pack, f"split_files/loras/{lora(stage)}", "loras")
    fetch("city96/umt5-xxl-encoder-gguf", ENCODER, "text_encoders")
    fetch("Comfy-Org/Wan-2.1", f"split_files/vae/{VAE}", "vae")

def expert(stage):
    return f"{CHECKPOINT}-{stage}Noise-{QUANT}.gguf"

def lora(stage):
    return f"wan2.2_i2v_lightx2v_4steps_lora_v1_{stage.lower()}_noise.safetensors"

def workflow(number, prefix):
    graph = {}
    def node(i, class_type, **inputs):
        graph[str(i)] = {"class_type": class_type, "inputs": inputs}
        return [str(i), 0]

    start, end = SHOTS[number]
    hi = node(1, "UnetLoaderGGUF", unet_name=expert("High"))
    lo = node(2, "UnetLoaderGGUF", unet_name=expert("Low"))
    hi = node(3, "LoraLoaderModelOnly", model=hi, lora_name=lora("High"), strength_model=1.0)
    lo = node(4, "LoraLoaderModelOnly", model=lo, lora_name=lora("Low"), strength_model=1.0)
    hi = node(5, "ModelSamplingSD3", model=hi, shift=5.0)
    lo = node(6, "ModelSamplingSD3", model=lo, shift=5.0)
    clip = node(7, "CLIPLoaderGGUF", clip_name=ENCODER, type="wan")
    vae = node(8, "VAELoader", vae_name=VAE)
    positive = node(9, "CLIPTextEncode", clip=clip, text=PROMPTS[number])
    negative = node(10, "CLIPTextEncode", clip=clip, text=NEGATIVE)
    first_image = node(11, "LoadImage", image=start)
    last_image = node(12, "LoadImage", image=end)
    node(13, "WanFirstLastFrameToVideo", positive=positive, negative=negative, vae=vae,
         width=WIDTH, height=HEIGHT, length=MODEL_FRAMES, batch_size=1,
         start_image=first_image, end_image=last_image)
    settings = dict(noise_seed=121000 + number, steps=4, cfg=1.0, sampler_name="euler",
                    scheduler="simple", positive=["13", 0], negative=["13", 1])
    sampled = node(14, "KSamplerAdvanced", model=hi, latent_image=["13", 2], add_noise="enable",
                   start_at_step=0, end_at_step=2, return_with_leftover_noise="enable", **settings)
    sampled = node(15, "KSamplerAdvanced", model=lo, latent_image=sampled, add_noise="disable",
                   start_at_step=2, end_at_step=4, return_with_leftover_noise="disable", **settings)
    video = node(16, "VAEDecodeTiled", samples=sampled, vae=vae, tile_size=256, overlap=64,
                 temporal_size=32, temporal_overlap=8)
    node(17, "SaveImage", images=video, filename_prefix=prefix)
    return graph

def start_comfy():
    server_log = (OUT / "GEMINUS_comfy.log").open("w")
    server = subprocess.Popen([sys.executable, "-u", "main.py", "--listen", "127.0.0.1", "--port", "8188",
                               "--input-directory", str(FRAMES), "--output-directory", str(SERVER_OUTPUT),
                               "--cuda-device", "0", "--lowvram", "--reserve-vram", "1.0",
                               "--fp16-unet", "--use-pytorch-cross-attention"],
                              cwd=COMFY, stdout=server_log, stderr=subprocess.STDOUT)
    server_log.close()
    for _ in range(180):
        if server.poll() is not None:
            raise RuntimeError((OUT / "GEMINUS_comfy.log").read_text()[-6000:])
        try:
            response = requests.get(f"{ROOT_URL}/object_info", timeout=3)
            response.raise_for_status()
        except requests.RequestException:
            time.sleep(2)
            continue
        missing = {item["class_type"] for item in workflow(1, "probe").values()} - response.json().keys()
        if missing:
            server.terminate()
            raise RuntimeError(f"ComfyUI nodes missing: {missing}. See GEMINUS_comfy.log.")
        return server
    server.terminate()
    raise TimeoutError("ComfyUI did not start within six minutes.")

def render(number, server):
    target = CLIPS / f"S{number:02d}.mp4"
    marker = target.with_suffix(".json")
    fingerprint = hashlib.sha256(json.dumps(workflow(number, "check"), sort_keys=True).encode())
    for name in SHOTS[number]:
        fingerprint.update((FRAMES / name).read_bytes())
    signature = fingerprint.hexdigest()
    if target.is_file() and marker.is_file() and marker.read_text() == signature:
        validate_clip(target, SHOT_SECONDS)
        log(f"S{number:02d}: already rendered")
        return target

    prefix = f"S{number:02d}_{time.time_ns()}"
    request = requests.post(f"{ROOT_URL}/prompt", json={"prompt": workflow(number, prefix)}, timeout=60)
    if not request.ok:
        raise RuntimeError(request.text[:3500])
    prompt_id = request.json()["prompt_id"]
    deadline = time.monotonic() + 45 * 60
    log(f"S{number:02d}: submitted to Wan 2.2")
    while time.monotonic() < deadline:
        if server.poll() is not None:
            raise RuntimeError("ComfyUI exited during render. See GEMINUS_comfy.log.")
        result = requests.get(f"{ROOT_URL}/history/{prompt_id}", timeout=45)
        result.raise_for_status()
        history = result.json().get(prompt_id)
        if history:
            status = history.get("status", {})
            if status.get("status_str") == "error":
                raise RuntimeError(str(status.get("messages", []))[-5500:])
            records = history.get("outputs", {}).get("17", {}).get("images", [])
            if not records:
                raise RuntimeError(f"S{number:02d}: no frames in completed output.")
            paths = [SERVER_OUTPUT / record["subfolder"] / record["filename"] for record in records]
            if len(paths) != MODEL_FRAMES or not all(path.is_file() for path in paths):
                raise RuntimeError(f"S{number:02d}: expected {MODEL_FRAMES} frames, got {len(paths)}.")
            break
        time.sleep(8)
    else:
        raise TimeoutError(f"S{number:02d} exceeded 45 minutes.")

    from tempfile import TemporaryDirectory
    count = FPS * SHOT_SECONDS
    with TemporaryDirectory(dir=TEMP) as directory:
        for index in range(count):
            source = paths[round(index * (len(paths) - 1) / (count - 1))]
            (Path(directory) / f"{index:03d}.png").symlink_to(source)
        pending = target.with_name(f"{target.stem}.pending.mp4")
        ffmpeg("-framerate", FPS, "-i", Path(directory) / "%03d.png", "-vf",
               "scale=720:1280:force_original_aspect_ratio=increase:flags=lanczos,crop=720:1280,setsar=1,format=yuv420p",
               "-frames:v", count, "-an", "-c:v", "libx264", "-preset", "fast", "-threads", "2",
               "-crf", "16", "-movflags", "+faststart", pending)
        validate_clip(pending, SHOT_SECONDS)
        pending.replace(target)
        marker.write_text(signature)
    for path in paths:
        path.unlink()
    log(f"S{number:02d}: 4.00 s exported")
    return target

def assemble(clips, output):
    listing = TEMP / "timeline.txt"
    listing.write_text("".join(f"file '{p.as_posix()}'\n" for p in clips))
    pending = output.with_name(f"{output.stem}.pending.mp4")
    ffmpeg("-f", "concat", "-safe", "0", "-i", listing, "-c", "copy", "-movflags", "+faststart", pending)
    validate_clip(pending, len(clips) * SHOT_SECONDS)
    pending.replace(output)

def main():
    if MODE not in {"sample", "full"}:
        raise ValueError("MODE must be 'sample' or 'full'.")
    unpack_assets(base64.b64decode("".join(ASSET_B64)))
    if not shutil.which("ffmpeg") or not shutil.which("ffprobe"):
        raise RuntimeError("ffmpeg and ffprobe are required.")
    prepare_comfy()
    server = start_comfy()
    try:
        numbers = [1] if MODE == "sample" else range(1, 13)
        clips = [render(number, server) for number in numbers]
        if MODE == "full":
            card = CLIPS / "S13.mp4"
            final_card(card)
            clips.append(card)
            output = OUT / "GEMINUS_Wan22_52s.mp4"
        else:
            output = OUT / "GEMINUS_Wan22_Test_04s.mp4"
        assemble(clips, output)
        log(f"READY: {output}")
    finally:
        server.terminate()
        try:
            server.wait(timeout=20)
        except subprocess.TimeoutExpired:
            server.kill()

main()
```
